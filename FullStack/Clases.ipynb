# Clase 1

## React

React es una libreria de JavaScript para construir interfaces de usuario a partir de componentes: funciones que reciben datos y devuelven la descripción de lo que se ve: 
    - Componente de React es una función de JS cuyo nombre empieza con mayúscula y que devuelve JSX
    - Vite es una herramienta de construccion para el frontend. Trae un servidor de desarrollo que entrega tus archivos al navegador, traduce el JSX a JS que el navegador entiende y       recarga la pagina cuando guardas.

## Pasos

1. Se realiza la instalación de las librerias desde la carpeta "pulsofit-fullstack" con "npm create vite@latest frontend -- --template react"
2. Se realiza la instalación de las carpetas desde la carpeta "frontend" con "cd frontend" para seleccionar la carpeta y "npm install" para descargas los respectivos archivos en dicha carpeta.
3. npm run dev para inicializar la pagina 

## JSX
Es una extensión de la sintaxis de JS que permite escribir etiquetas parecidas a HTML dentro del código. No es HTML ni un texto: cada etiqueta se traduce a una llamada de JS que describe un elemento.

## Tailwind CSS
Es un framework de CSS utility-first: trae miles de clases pequeñas, cada una con una o dos declaraciones de CSS, que se combinan directamente en el atributo className.

## Props
Las props (de properties) son los datos que un componente recibe del componente que lo usa. Se escriben como atributos en la etiqueta (<TarjetaClase clase={…} />) y llegan a la función en un objeto.
