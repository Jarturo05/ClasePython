## Como funciona CSS:
Que es una regla:
    - 

* Selector
    1. Selector de etiqueta: Se llama a sus etiquetas que se quiera cambiar.
    2. Selector de clase: Nombres propios a cada elemento donde se pueda cambiar especificamente.
    3. Selector Universal: Absolutamente todos.
    4. Selector de ID

* Modelo de caja
    1. Contenido: El texto de la clase
    2. Relleno: Padding
    3. Border: El marco
    4. Margin: Espacio afuera

* Repasar Flexbox, modelo de caja y como se alinean los textos
    - Flexbox se activa con el padre como: display: flex
    - Tailwind: Coleccion de clases: Nos va a ayudar para que al hacerlo en CSS sea mas "fácil" como bootstrap tambien

* Google fonts: Para traer fuentes de letras

## Como funciona JavaScript:
- DOM: (Document Object Model): Es la interpretacion en el codigo de los componentes del HTML, donde se efectuan los eventos o cambios actuales de la página, donde javascript intenta  
entender o interpetrar a HTML.
