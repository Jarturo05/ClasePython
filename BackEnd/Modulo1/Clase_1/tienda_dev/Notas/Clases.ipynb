# Notas Clase 1

Api Restfull --> Api basadas en Http es decir basadas en Get, post, put y delete

Endpoint: Una URL especifica de tu API que hace algo, como http://miapi.com/productos...etc

HTTP: El idioma de internet como son los verbos que voy a utilizar
Los verbos son:
* Get: Pedir u obtener la info de un producto
* Post: Crear o registrar un nuevo producto
* Put: Actualizar un producto (Casi siempre se usa por ID)
* Delete: Eliminar un producto

CRUD: Create, Read, Update and Delete

Rest(Representational State Transfer): Las reglas que lo ordenan
JSON(Javascript Object Notation) es texto estructurado con llaves {} y pares clave-valor: El formato de los datos que trae al usuario, texto plano

## Regla 1
Cada recurso tiene una URL única, cada espacio es un "endpoint"

## Regla 2
Usas verbos HTTP para indicar la acción.

## Instalación de la API

1. Instalación respectiva de Python
2. Crear carpeta del Proyecto
3. Creación del entorno virtual (venv)
    * mk dir "tienda_dev" (Esto dentro de la terminal)
    * cd tienda_dev (Para trabajar en la carpeta "tienda_dev")
    * Primero se selecciona Python + Shift + P para ir a Python: Select Interpreter, se selecciona la version de python que termine en .exe
    * python -m venv venv (Para crear el entorno virtual)
    * .\venv\Scripts\Activate.ps1 o en su defecto \Activate (Para activar el entorno)
4. Creacion del Framework
    * píp install fastapi uvicorn (Instalacion de el Framework, uvicorn es el servidor que va a funcionar como app web)
    * mkdir app (Se crea la carpeta app)
    * mkdir app\routers (Creacion de las carpetas routers, services, models, schemas, config)
        * \routers: Todas las rutas(endpoints) agrupados por funcionalidad.
        * \services: Logica de negocio o funciones que hacen las cosas.
        * \models: Modelado de la base de datos
        * \Schemas: Define los datos que se esperan recibir o resolver de los endpoints
        * \congif: Todo el tema de configuraciones muy precisas.
    * ni app\__init__.py (Esto crea el package donde se realiza en cada carpeta antes creada para conectar las carpetas y conectarlas)
5. Iniciar el proyecto web con app/main.py
6. Levantar o inicializar la app uvicorn app.main:app --reload
7. Iniciar app
    * cd Clase_1\tienda_dev
    * venv\Scripts\activate
    * uvicorn app.main:app --reload

# Clase 3

Arquitectura por capas(Capas profesionales): Organizar la app en sus respectivas carpetas antes nombradas

## Separar por responsabilidades:

    * Capa de rutas (router): recibe la petición HTTP y devuelve la respuesta. Es la "recepción"
    * Capa de servicio (service): contiene la lógica de negocio (cómo se busca, se crea, se borra). Es la cocina
    * Capa de datos: Guarda la información. Es la bódega.

    A parte los schemas (validación) viven en su propia carpeta, porque validar tambieén es un trabajo propio.

- Siempre se manje la capa ruta (router) se comunica con services y services con los datos

APIRouter: es una clase de FastAPI que funciona como una mini_aplicación de rutas: permite definir endpoints en un archivo aparte, agrupados para luego conectarlos a la app principal con app.include_router(). Es el mecanismo oficial de FastAPI para dividir una API en múltiples archivos.
